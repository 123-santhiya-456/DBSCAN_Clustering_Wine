```python
# DBSCAN Clustering on Wine Dataset

import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import numpy as np
```

```python
# Load dataset
wine = load_wine()
X = wine.data
```

```python
# Feature scaling (VERY IMPORTANT for DBSCAN)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Apply DBSCAN
dbscan = DBSCAN(eps=1.2, min_samples=5)
labels = dbscan.fit_predict(X_scaled)
```

```python
# Number of clusters (excluding noise)
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
print("Number of clusters:", n_clusters)
```

```python
# Silhouette Score (only if more than 1 cluster)
if n_clusters > 1:
    score = silhouette_score(X_scaled, labels)
    print("Silhouette Score:", score)
else:
    print("Silhouette Score not applicable")
```

```python
# Visualization (first two features)
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("DBSCAN Clustering on Wine Dataset")
plt.show()
```
